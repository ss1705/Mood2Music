```python
from src.retrieval.retriever import (
    load_catalog,
    load_2a_artifacts,
    search_regulation
)
```

```python
catalog = load_catalog()
model_2a, index_2a, map_2a = load_2a_artifacts()
print(f"Catalog: {len(catalog)} tracks")
print("Phase 2A artifacts loaded")
```

```python
test_pairs = [
    ("I feel stressed and overwhelmed",     "I want to feel calm and at peace"),
    ("I feel low energy and unmotivated",   "I want to feel energized and ready to go"),
    ("I feel anxious and restless",         "I want to feel focused and grounded"),
]
```

```python
for current_mood, target_mood in test_pairs:
    result = search_regulation(
        current_mood, target_mood,
        model_2a, index_2a, map_2a,
        catalog,
        n_waypoints=5,
        tracks_per_waypoint=2
    )

    print(f"\n{'='*70}")
    print(f"FROM: '{current_mood}'")
    print(f"  TO: '{target_mood}'")
    print(f"{'='*70}")
    print(f"Current coords: valence={result['current_coords'][0]:.3f}, "
          f"energy={result['current_coords'][1]:.3f}")
    print(f"Target  coords: valence={result['target_coords'][0]:.3f}, "
          f"energy={result['target_coords'][1]:.3f}")
    print()

    for wp in result['waypoints']:
        print(f"  Step {wp['step']+1}/5  "
              f"(val={wp['valence']:.3f}, nrg={wp['energy']:.3f})")
        for track in wp['tracks']:
            print(f"    → {track['title'][:40]:<40} "
                  f"val={track['valence']:.2f} "
                  f"nrg={track['energy']:.2f} "
                  f"dist={track['va_distance']:.3f}")
        print()
```
